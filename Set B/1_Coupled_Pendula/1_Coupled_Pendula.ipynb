## Curve fitting for exponential decay

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# --- Fit function ---
def damped_sine(t, A, gamma, omega, phi, C):
    return A * np.exp(-gamma * t) * np.sin(omega * t + phi) + C

# --- Example: fit P1 data ---
# (replace these with your real LabQuest data arrays)
t = time1

y1 = []
for i in range(len(tn_s1)):
    y1.append(np.mean([tn_s1[i], tn_s2[i]]))
y1 = np.array(y1)
   
# Initial parameter guesses
A0 = (max(y1) - min(y1)) / 2
gamma0 = 0.05
omega0 = 2 * np.pi / 2.2   # roughly 2.2 s period, adjust if needed
phi0 = 0
C0 = 0

popt, pcov = curve_fit(damped_sine, t, y1, p0=[A0, gamma0, omega0, phi0, C0])
A, gamma, omega, phi, C = popt

print(f"Fitted parameters:\nA={A:.4f}, gamma={gamma:.4f}, ω={omega:.4f}, φ={phi:.4f}, C={C:.4f}")

# --- Generate fitted curve and envelope ---
y_fit = damped_sine(t, *popt)
envelope = np.abs(A * np.exp(-gamma * t))

# --- Plot ---
plt.figure(figsize=(10,4))

# plt.plot(t, y1, label='Experimental (P1)', alpha=0.7)
# plt.plot(t, y_fit, 'r', label='Fitted Damped Sine')

plt.plot(time1, tn_s1, label='P1')
plt.plot(time1, tn_s2, label='P2')

plt.plot(t, envelope, 'k--', alpha=0.5, label='Envelope (+)')
plt.plot(t, -envelope, 'k--', alpha=0.5, label='Envelope (–)')

plt.xlabel('Time (s)', fontsize=13)
plt.ylabel('Position (m)', fontsize=13)
plt.title('In-phase pendula (mean of 3 experimental trials) \n with fitted decay functions', fontsize=14)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Constants
g = 9.81             # m/s^2
L = 1.02             # m
m1 = 4.334           # kg
m2 = 4.014           # kg
k = 26.5             # N/m (from your spring experiment)

# Derived parameters (approx average mass)
m = (m1 + m2) / 2

# Angular frequencies
omega0 = np.sqrt(g / L)
omega_c = np.sqrt(k / m)

omega_plus = np.sqrt(omega0**2 + omega_c**2)   # out-of-phase
omega_minus = np.sqrt(omega0**2 - omega_c**2)  # in-phase
omega_beat = omega_plus - omega_minus
T_beat = 2 * np.pi / omega_beat

print(f"ω₀ = {omega0:.3f} rad/s")
print(f"ω₊ = {omega_plus:.3f} rad/s, ω₋ = {omega_minus:.3f} rad/s")
print(f"Beat period ≈ {T_beat:.2f} s")


# Initial angle (15 degrees)
theta0 = np.deg2rad(15)

# Time array
t = np.linspace(0, 30, 2000)

# Angular displacements
theta1 = theta0 * np.cos((omega_plus + omega_minus)/2 * t) * np.cos((omega_plus - omega_minus)/2 * t)
theta2 = theta0 * np.sin((omega_plus + omega_minus)/2 * t) * np.sin((omega_plus - omega_minus)/2 * t)

# Convert to approximate position (small-angle: x ≈ L * θ)
x1 = L * theta1
x2 = L * theta2

# Envelope
envelope = theta0 * np.cos((omega_plus - omega_minus)/2 * t)

# Plot
plt.figure(figsize=(15,5))
plt.plot(t, x1, label='Pendulum 1')
plt.plot(t, x2, label='Pendulum 2')
plt.plot(t, L*envelope, 'k--', alpha=0.4, label='Envelope (+)')
plt.plot(t, -L*envelope, 'k--', alpha=0.4, label='Envelope (–)')

plt.title('Theoretical Coupled Pendula Motion and Beat Envelope')
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import hilbert, welch, detrend, windows
from scipy.fft import rfft, rfftfreq

# -------------------------
# Replace these with your actual arrays or CSV load
# t = ...
# p1 = ...
# p2 = ...
# Example: load from CSV with columns time,p1,p2:
# import pandas as pd
# df = pd.read_csv('yourfile.csv')
# t = df['time'].values
# p1 = df['p1'].values
# p2 = df['p2'].values
# -------------------------

# quick checks
t = np.asarray(time1)
p1 = np.asarray(t1_s1)
p2 = np.asarray(t1_s2)
N = len(t)
T_total = t[-1] - t[0]
dt = np.mean(np.diff(t))
fs = 1.0/dt
print(f"Samples={N}, duration={T_total:.2f}s, fs={fs:.2f} Hz, df={1/T_total:.4f} Hz")

# --- preprocessing ---
# remove mean & linear trend
p1c = detrend(p1, type='linear')
p2c = detrend(p2, type='linear')

# optional: apply a Hann window before FFT to reduce spectral leakage
win = windows.hann(N)
p1w = p1c * win
p2w = p2c * win

# -------------------------
# FFT of raw signals (single-shot)
# -------------------------
Y1 = rfft(p1w)
Y2 = rfft(p2w)
freqs = rfftfreq(N, dt)

# amplitude spectra (correct for window energy)
# normalization so amplitude corresponds roughly to signal amplitude
norm = np.sum(win)/N
amp1 = np.abs(Y1) / (N*norm)
amp2 = np.abs(Y2) / (N*norm)

# -------------------------
# Better PSD estimate: Welch (averaged) for improved SNR
# -------------------------
# choose segment length for welch: e.g., 4-second segments (tune if needed)
nperseg = int(round(4.0 * fs))
if nperseg >= N:
    nperseg = N//2
f_w, Pxx1 = welch(p1c, fs=fs, window='hann', nperseg=nperseg, noverlap=nperseg//2)
_,      Pxx2 = welch(p2c, fs=fs, window='hann', nperseg=nperseg, noverlap=nperseg//2)

# convert PSD to amplitude spectral density-ish (sqrt(PSD)) for plotting
ASD1 = np.sqrt(Pxx1)
ASD2 = np.sqrt(Pxx2)

# -------------------------
# Envelope (Hilbert) and PSD of envelope (to reveal slow beat freq)
# -------------------------
env1 = np.abs(hilbert(p1c))
env2 = np.abs(hilbert(p2c))

# optionally lowpass the envelopes or smooth them if very noisy (simple moving avg)
def moving_average(x, window_samples):
    if window_samples < 2:
        return x
    w = np.ones(window_samples)/window_samples
    return np.convolve(x, w, mode='same')

env1s = moving_average(env1, max(3, int(round(0.05 * fs))))  # e.g., 0.05s smoothing
env2s = moving_average(env2, max(3, int(round(0.05 * fs))))

# Welch on envelope to find beat frequencies (use long segments because beat is low freq)
nperseg_env = int(round(8.0 * fs))
if nperseg_env >= N:
    nperseg_env = N//2
f_env, Penv1 = welch(env1s - np.mean(env1s), fs=fs, window='hann', nperseg=nperseg_env, noverlap=nperseg_env//2)
_,      Penv2 = welch(env2s - np.mean(env2s), fs=fs, window='hann', nperseg=nperseg_env, noverlap=nperseg_env//2)

ASD_env1 = np.sqrt(Penv1)
ASD_env2 = np.sqrt(Penv2)

# -------------------------
# Theoretical mode frequencies (from your parameters)
# -------------------------
g = 9.81
L = 1.02
k = 26.5
m1 = 4.334
m2 = 4.014
m_avg = (m1 + m2)/2
omega0 = np.sqrt(g/L)
omega_c = np.sqrt(k/m_avg)
# For non-identical masses a 2x2 eigenvalue solve is better; here we print approx:
omega_plus = np.sqrt(omega0**2 + omega_c**2)
omega_minus = np.sqrt(max(omega0**2 - omega_c**2, 0.0))
f_plus = omega_plus/(2*np.pi)
f_minus = omega_minus/(2*np.pi)
print(f"theoretical f_plus = {f_plus:.3f} Hz, f_minus = {f_minus:.3f} Hz, Δf = {abs(f_plus-f_minus):.3f} Hz")

# -------------------------
# Plot: time signals (small), spectra, and envelope PSD
# -------------------------
plt.figure(figsize=(12,9))

plt.subplot(3,1,1)
plt.plot(t, p1c, label='P1 (detrended)', alpha=0.8)
plt.plot(t, p2c, label='P2 (detrended)', alpha=0.8)
plt.plot(t, env1s, '--', label='env1 (smoothed)', alpha=0.9)
plt.plot(t, env2s, '--', label='env2 (smoothed)', alpha=0.9)
plt.xlim(t[0], t[-1])
plt.ylabel('Position (m)')
plt.legend()
plt.title('Time signals and envelopes')

plt.subplot(3,1,2)
plt.semilogy(freqs, amp1, label='FFT P1 (windowed)', alpha=0.6)
plt.semilogy(freqs, amp2, label='FFT P2 (windowed)', alpha=0.6)
plt.semilogy(f_w, ASD1, label='Welch P1 (ASD)', linewidth=2)
plt.semilogy(f_w, ASD2, label='Welch P2 (ASD)', linewidth=2)
plt.axvline(f_plus, color='k', linestyle='--', label=f'f_plus={f_plus:.3f}Hz')
plt.axvline(f_minus, color='k', linestyle=':', label=f'f_minus={f_minus:.3f}Hz')
plt.xlim(0, max(2.0, f_plus*3))
plt.xlabel('Frequency (Hz)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True, which='both', ls='--', alpha=0.4)

plt.subplot(3,1,3)
plt.semilogy(f_env, ASD_env1, label='Env PSD P1')
plt.semilogy(f_env, ASD_env2, label='Env PSD P2')
plt.xlim(0, 1.0)   # beats are low-frequency — show up below ~1 Hz
plt.xlabel('Frequency (Hz)')
plt.ylabel('Envelope amplitude (AU)')
plt.legend()
plt.grid(True, which='both', ls='--', alpha=0.4)

plt.tight_layout()
plt.show()

# -------------------------
# Diagnostics & remarks
# -------------------------
# Print frequency resolution and ability to resolve peaks
df = 1.0 / T_total
print(f"frequency resolution df = {df:.4f} Hz")
print("If two peaks are closer than ~df (or if SNR is low), they will be hard to resolve.")
```
