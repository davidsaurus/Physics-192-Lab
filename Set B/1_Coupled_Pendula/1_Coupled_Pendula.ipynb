## Importing packages/libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
coupled1 = pd.read_csv('data/coupled1.csv')
coupled2 = pd.read_csv('data/coupled2.csv')
coupled3 = pd.read_csv('data/coupled3.csv')

inphase1 = pd.read_csv('data/inphase1.csv')
inphase2 = pd.read_csv('data/inphase2.csv')
inphase3 = pd.read_csv('data/inphase3.csv')

p1_single = pd.read_csv('data/p1_single.csv')
p2_single = pd.read_csv('data/p2_single.csv')

resonanceExtra = pd.read_csv('data/resonanceExtra.csv')
```

## Resonance

```python
time1 = coupled3['Run 3:Time(s)']
time2 = coupled3['Run 4:Time(s)']
time3 = coupled3['Run 5:Time(s)']
time4 = coupled3['Run 6:Time(s)']

### Naming scheme:
## tn = trial n, n=1,2,3
## sn = sensor m, m = 1,2.

equi_point = 0.275 # meters

t1_s1 = coupled3['Run 3:Position(m)']-equi_point
t1_s2 = -(coupled3['Run 3:Position 1(m)']-equi_point)

t2_s1 = coupled3['Run 4:Position(m)']-equi_point
t2_s2 = -(coupled3['Run 4:Position 1(m)']-equi_point)

t3_s1 = coupled3['Run 5:Position(m)']-equi_point
t3_s2 = -(coupled3['Run 5:Position 1(m)']-equi_point)

t4_s1 = coupled3['Run 6:Position(m)']-equi_point
t4_s2 = -(coupled3['Run 6:Position 1(m)']-equi_point)
```

```python
plt.figure(figsize = (8, 4))

plt.title('Coupled pendula trial 1', fontsize=20)
plt.xlabel('Time (s)', fontsize=15)
plt.ylabel('Position (m)', fontsize=15)

plt.plot(time1, t1_s1, label='P1')
plt.plot(time1, t1_s2, label='P2')

plt.xticks()
#plt.yticks(np.linspace(np.min(t1_s1), np.max(t1_s2), 6))

plt.grid()
plt.legend()

plt.show()
```

```python
plt.figure(figsize = (8, 4))

plt.title('Coupled pendula trial 2', fontsize=20)
plt.xlabel('Time (s)', fontsize=15)
plt.ylabel('Position (m)', fontsize=15)

plt.plot(time2, t2_s1, label='P1')
plt.plot(time2, t2_s2, label='P2')

plt.xticks()
plt.yticks(np.linspace(np.min(t2_s2), np.max(t2_s2), 6))

plt.grid()
plt.legend()

plt.show()
```

```python
plt.figure(figsize = (8, 4))

plt.title('Coupled pendula trial 1', fontsize=20)
plt.xlabel('Time (s)', fontsize=15)
plt.ylabel('Position (m)', fontsize=15)

plt.plot(time3, t3_s1, label='P1')
plt.plot(time3, t3_s2, label='P2')

plt.xticks()
plt.yticks()

plt.grid()
plt.legend()

plt.show()
```

```python
plt.figure(figsize = (8, 4))

plt.title('Coupled pendula trial 4', fontsize=20)
plt.xlabel('Time (s)', fontsize=15)
plt.ylabel('Position (m)', fontsize=15)

plt.plot(time4, t4_s1, label='P1')
plt.plot(time4, t4_s2, label='P2')

plt.xticks()
plt.yticks()

plt.grid()
plt.legend()

plt.show()
```
