```python
coupled1 = pd.read_csv('data/coupled1.csv')
coupled2 = pd.read_csv('data/coupled2.csv')
coupled3 = pd.read_csv('data/coupled3.csv')

inphase1 = pd.read_csv('data/inphase1.csv')
inphase2 = pd.read_csv('data/inphase2.csv')
inphase3 = pd.read_csv('data/inphase3.csv')

p1_single = pd.read_csv('data/p1_single.csv')
p2_single = pd.read_csv('data/p2_single.csv')

resonanceExtra = pd.read_csv('data/resonanceExtra.csv')
```

```python
coupled3 = coupled3.iloc[:, 14:42]
```

```python
time1 = coupled3.iloc[:, 0]
time2 = coupled3.iloc[:, 7]
time3 = coupled3.iloc[:, 14]
time4 = coupled3.iloc[:, 21]

### Naming scheme:
## tn = trial n, n=1,2,3
## sn = sensor m, m = 1,2.

t1_s1 = coupled3.iloc[:, 1]
t1_s2 = coupled3.iloc[:, 4]

t2_s1 = coupled3.iloc[:, 8]
t2_s2 = coupled3.iloc[:, 11]

t3_s1 = coupled3.iloc[:, 15]
t3_s2 = coupled3.iloc[:, 18]

t4_s1 = coupled3.iloc[:, 22]
t4_s2 = coupled3.iloc[:, 25]
```

```python
plt.figure(figsize = (8, 4))

plt.title('Coupled pendula trial 1', fontsize=20)
plt.xlabel('Time (s)', fontsize=15)
plt.ylabel('Position (m)', fontsize=15)

plt.plot(time1, t1_s1, label='P1')
plt.plot(time1, t1_s2, label='P2')

plt.xticks()
plt.yticks(np.linspace(np.min(t1_s1), np.max(t1_s2), 6))

plt.grid()
plt.legend()

plt.show()
```
