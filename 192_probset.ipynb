## Preliminaries

```python
import numpy as np
import scipy.special as sp
import scipy.stats as st
```

## Defining functions

```python
def CDF(t, v):
    u = (v+1)*0.5
    return 0.5 + t*sp.gamma(u) * sp.hyp2f1(0.5, u, 1.5, (-t**2)*0.5) / (np.sqrt(np.pi*v)*sp.gamma(v*0.5))

def p_value(t, v):
    return 2 * (1-CDF(t, v))
```
