```python
import numpy as np
import scipy.special as sp
import scipy.stats as st
```

```python
def CDF(t, v):
    return 0.5 + ( t*sp.gamma((v+1)*0.5) * sp.hyp2f1(0.5, (v+1)*0.5, 1.5, -t**2 * 0.5)/(np.sqrt(np.pi*v)*sp.gamma(v*0.5)) )

def p_value(t, v):
    return 2 * (1-CDF(t, v))
```
