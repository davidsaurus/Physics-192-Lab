```python
### Import packages

import numpy as np
import pandas as pd
import scipy.special as sp
```

## Problem 1

```python
### Data and metrics

data = {
        'mat' : ['Cu', 'Al', 'Ni', 'Fe', 'Brass', 'Ag', 'W', 'Constantan'],
        'no_field' : [10.12, 10.45, 12.30, 15.20, 11.10, 9.85, 14.75, 13.60],
        'w_field' : [10.18, 10.50, 12.42, 15.35, 11.22, 9.90, 14.89, 13.71]
    }

df1 = pd.DataFrame(data)
df1 = df1.set_index('mat')
df1['diff'] = np.abs(df1['no_field'] - df1['w_field'])

diff_mean = df1['diff'].mean()
diff_std = df1['diff'].std()

metrics = {
        'diff_mean' : [diff_mean], 
        'diff_std' : [diff_std],
    }

df2 = pd.DataFrame(metrics)
```

```python
### Calculate t-statistic

def deg_f_paired(N):
    return N-1

dof_paired = deg_f_paired(df1.shape[0])

def paired_t_test(N, d, d0, s):
    return (d-d0) / (s / np.sqrt(N))

t_paired = paired_t_test(df1.shape[0], diff_mean, 0, diff_std)

### Calculate p-value

def CDF(t, v):
    u = (v+1)*0.5
    return 0.5 + t*sp.gamma(u) * sp.hyp2f1(0.5, u, 1.5, (-t**2)/v) / \
        (np.sqrt(np.pi*v)*sp.gamma(v*0.5))

def p_value(t, v):
    return 2 * (1-CDF(t, v))

p_val1 = p_value(t_paired, dof_paired)
```

```python
### Results

print('Relevant metrics: \n', df2, '\n')

print('################################# \n')

print(f't-value: {t_paired} \n')

print('################################# \n')

print(f'p-value: {p_val1}')
print(f'p-value (in %): {round(p_val1*100, 5)} %')

print(f'Deg freedom paired: {dof_paired}')
```

## Problem 2

```python
### Data and metrics

data = {
        'las_a' : [4.85, 4.92, 4.88, 4.91, 4.89, 4.93],
        'las_b' : [4.75, 4.70, 4.78, 4.73, 4.80, 4.76],
    }

df3 = pd.DataFrame(data)

a_mean = df3['las_a'].mean()
b_mean = df3['las_b'].mean()

a_var = df3['las_a'].var()
b_var = df3['las_b'].var()

metrics = {
        'a_mean' : [a_mean], 
        'b_mean' : [b_mean],
        'a_var': [a_var],
        'b_var': [b_var],
        'a_std': [np.sqrt(a_var)],
        'b_std': [np.sqrt(b_var)]
    }

df4 = pd.DataFrame(metrics)
```

```python
### Calculate t-statistic

def deg_f_unpaired(N, M, n_var, m_var):
    return ( ((n_var/N) + (m_var)/M) )**2 / ( (n_var / (N**2) * (N-1)) * (m_var / (M**2) * (M-1)) )

dof_unpaired = deg_f_unpaired(len(df3['las_a']), len(df3['las_b']), a_var, b_var)

def unpaired_t_test(N, M, n_mean, m_mean, n_var, m_var):
    return (n_mean - m_mean) / np.sqrt((n_var/N) + (m_var/M))

t_unpaired = unpaired_t_test(len(df3['las_a']), len(df3['las_b']), a_mean, b_mean, a_var, b_var)

### Calculate p-value

def CDF(t, v):
    u = (v+1)*0.5
    return 0.5 + t*sp.gamma(u) * sp.hyp2f1(0.5, u, 1.5, (-t**2)/v) / \
        (np.sqrt(np.pi*v)*sp.gamma(v*0.5))

def p_value(t, v):
    return 2 * (1-CDF(np.abs(t), v))

p_val2 = p_value(t_unpaired, dof_unpaired)
```

```python
### Results

print('Relevant metrics: \n', df4, '\n')

print('################################# \n')

print(f't-value: {t_unpaired} \n')

print('################################# \n')

print(f'p-value: {p_val2}')
print(f'p-value (in %): {round(p_val2*100, 5)} %')

print(f'Deg freedom unpaired: {dof_unpaired}')
```

## Problem 3

```python
### Data and metrics

data = {
        'A' : [0.48, 0.52, 0.50],
        'B' : [0.32, 0.36, 0.34],
        'C' : [0.10, 0.12, 0.09]
    }

df5 = pd.DataFrame(data)

A_mean = df5['A'].mean()
B_mean = df5['B'].mean()
C_mean = df5['C'].mean()

A_var = df5['A'].var()
B_var = df5['B'].var()
C_var = df5['C'].var()

G_mean = df5.mean(axis=None)

metrics = {
        'A_mean' : [A_mean], 
        'B_mean' : [B_mean],
        'C_mean' : [C_mean],
        'G_mean' : [G_mean],
        'A_var': [A_var],
        'B_var': [B_var],
        'C_var': [C_var]
    }

df6 = pd.DataFrame(metrics)
```

```python
### Calculate deg freedom

def deg_bet(k):
    return k-1

def deg_with(N, k):
    return N-k

dof_bet = deg_bet(df5.shape[0])
dof_with = deg_with(df5.shape[0]*df5.shape[1], df5.shape[0])
    
### Calculate SS

def SSB(xi, xg):
    squares = [3*(xi[i] - xg)**2 for i in range(len(xi))]
    return np.sum(squares)

ssb = SSB([A_mean, B_mean, C_mean], G_mean)

def SSW(xij, xi):
    xij = xij.to_numpy()
    square = []
    for i in range(0, xij.shape[1]):
        square.append(np.sum((xij[:, i] - xi[i])**2))
    return np.sum(square)
    
ssw = SSW(df5, [A_mean, B_mean, C_mean])
```

```python
### Calculate MS and F-stat

def MSB(ssb, dfb):
    return ssb/dfb

def MSW(ssw, dfw):
    return ssw/dfw

msb = MSB(ssb, dof_bet)
msw = MSW(ssw, dof_with)

def F_stat(msb, msw):
    return msb/msw

F = F_stat(msb, msw)

### Calculate p-value
def p_value_f(d1, d2, F):
    CDF = sp.betainc(d1*0.5, d2*0.5, d1*F/((d1*F)+d2))
    return 1-CDF

p_val3 = p_value_f(dof_bet, dof_with, F)
```

```python
### Results

print('Relevant metrics: \n', df6, '\n')

print('################################# \n')

print('SS:', ssb, ssw, '\n', 'MS:', msb, msw)
print(f'F-value: {F} \n')

print('################################# \n')

print(f'p-value: {p_val3}')
print(f'p-value (in %): {round(p_val3*100, 5)} %')

print(f'Degs freedom: {dof_bet, dof_with}')
```
