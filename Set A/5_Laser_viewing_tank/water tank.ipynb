# Pixel Coordinates

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def locate_pixel(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDBLCLK:
        #img_copy = img.copy()
        cv2.rectangle(img_copy, (0, 0), (180, 40), (0, 0, 0), -1)
        cv2.putText(img_copy, f'({x}, {img_y - y})', (0, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2, lineType=cv2.LINE_AA)
        coords.append([x, img_y - y])

def calculate_angle(coords):
    coords = np.array(coords)

    vertical1 = np.array([0, 1])
    vertical2 = np.array([0, -1])

    vec1 = coords[0] - coords[1]
    vec2 = coords[2] - coords[1]

    dot_prod1 = np.dot(vec1, vertical1)
    dot_prod2 = np.dot(vec2, vertical2)

    vec1 = np.linalg.norm(vec1)
    vec2 = np.linalg.norm(vec2)
    vertical1_mag = np.linalg.norm(vertical1)
    vertical2_mag = np.linalg.norm(vertical2)

    angle1 = np.arccos(dot_prod1 / (vec1 * vertical1_mag))
    angle1 = np.degrees(angle1)

    angle2 = np.arccos(dot_prod2 / (vec2 * vertical2_mag))
    angle2 = np.degrees(angle2)

    return angle1, angle2
```

# Load Images

```python
img = cv2.imread("images/refraction_1.jpg")
img = cv2.resize(img, None, fx=0.7, fy=0.7)
img_y, img_x, _ = img.shape

img_copy = img.copy()

plt.imshow(img)
plt.show()

windowName = "Water Tank"

coords = []

cv2.namedWindow(windowName)
cv2.setMouseCallback(windowName, locate_pixel)

while True:
    cv2.imshow(windowName, img_copy)
    if cv2.waitKey(20) & 0xFF == 27:
        break

cv2.destroyAllWindows()

if len(coords) >= 3:
    angle1, angle2 = calculate_angle(coords)
    print(angle1, "degrees")
    print(angle2, "degrees")

    print(np.sin(np.radians(angle1)) / np.sin(np.radians(angle2)))
```
