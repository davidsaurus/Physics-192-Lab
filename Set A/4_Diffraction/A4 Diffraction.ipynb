# Measurement of Wavelength / Grazing Incidence ()

$$\begin{align}
\lambda n = d (\sin \theta_i - \sin \theta_r)
\end{align}$$
where:
$$\begin{align}
\theta_i = 90^o - \frac{\Psi_0}{2} &&, \theta_r = 90^o + \frac{\Psi_0}{2} &&, \tan\Psi_0 = \frac{h_0}{L} &&, \tan\Psi_0 = \frac{h_0}{L}
\end{align}$$
where:
* $L$ the distance from the ruler to the screen.
* $h_0$ is the height from the laser's dot on the screen without the ruler.

Common wavelength of HeNe red laser: 632.8 nm or 6.328e-5 cm

```python
def calc_wavelength(n, d, S_n, S_0, L):
    psi_0 = np.arctan(S_0 / L)
    psi_n = np.arctan(S_n / L)
    theta_i = np.pi/2 - psi_0/2
    theta_r = np.pi/2 - psi_n + psi_0/2
    return d/n * (np.sin(theta_i) - np.sin(theta_r))
    #return d / (2*(L**2)*n) * (S_n**2 - S_n * S_0)

d = 0.05 # cm
L = 64.8 # cm
h_0 = 14.0

print(calc_wavelength(1, d, 14.8, h_0, L))
print(calc_wavelength(2, d, 15.5, h_0, L))
print(calc_wavelength(3, d, 16.1, h_0, L))
print(calc_wavelength(4, d, 16.8, h_0, L))
print(calc_wavelength(5, d, 17.4, h_0, L))
print(calc_wavelength(6, d, 17.9, h_0, L))
```

# Diffraction
$$
d = \frac{m\lambda}{\sin\theta}
$$
where:
* $m$ is the order number for the bright fringes
* $\lambda$ the wavelength of light
* $\theta$ the angle of diffraction
* $d$ the distance between slits

## Angle of Diffraction
$$
\theta = \tan^{-1}(\frac{x}{D})
$$
where $D$ is the distance from the diffracting grating and screen.

```python
def calc_d(m, wavelength, x, D):
    theta = np.arctan(x/D)
    return (m*wavelength) / np.sin(theta)

# CD
wavelength = 6.32e-5    # empirical wavelength of HeNe red lasers

print(calc_d(1, wavelength, 56.3/2, 60.6))
print(calc_d(1, wavelength, 11/2, 12))
print(calc_d(2, wavelength, 38/2, 12))

# DVD
wavelength = 6.32e-5    # empirical wavelength of HeNe red lasers

print(calc_d(1, wavelength, 70.5/2, 21))
print(calc_d(1, wavelength, 38/2, 11.5))
```
